## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import pandas as pd
import numpy as np
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_12_dataset.csv")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
print("Shape of dataset:", df.shape)
print("Column names:", list(df.columns))
df.info()
print("\\nMissing values:")
display(df.isnull().sum())
print("Number of duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
quantity_array = df["Quantity"].to_numpy()
price_array = df["Unit_Price"].to_numpy()
average_quantity = np.mean(quantity_array)
highest_unit_price = np.max(price_array)
lowest_unit_price = np.min(price_array)
print("Average quantity per record:", round(average_quantity, 2))
print("Highest unit price:", highest_unit_price)
print("Lowest unit price:", lowest_unit_price)
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
analysis_df = df.copy()

analysis_df["Revenue"] = analysis_df["Quantity"] * analysis_df["Unit_Price"]

analysis_df["Price_Category"] = np.where(
    analysis_df["Unit_Price"] >= 30,
    "High Price",
    "Regular Price"
)

display(analysis_df.head())
```

## 6. Your individual project challenge
**Challenge:** Calculate revenue per row; find best-selling item by quantity and highest-revenue day.

Write your solution below and explain what the result means.


```python
item_sales = analysis_df.groupby("Item")["Quantity"].sum()
item_sales = item_sales.sort_values(ascending=False)

print("Total quantity sold by item:")
display(item_sales)
best_selling_item = item_sales.idxmax()
best_selling_quantity = item_sales.max()

print("Best-selling item:", best_selling_item)
print("Total quantity sold:", best_selling_quantity)

daily_revenue = analysis_df.groupby("Day")["Revenue"].sum()
daily_revenue = daily_revenue.sort_values(ascending=False)

print("\\nTotal revenue by day:")
display(daily_revenue)

highest_revenue_day = daily_revenue.idxmax()
highest_day_revenue = daily_revenue.max()

print("Highest-revenue day:", highest_revenue_day)
print("Revenue on that day:", highest_day_revenue)

total_revenue = np.sum(analysis_df["Revenue"].to_numpy())
average_revenue = np.mean(analysis_df["Revenue"].to_numpy())

print("\\nTotal revenue from all records:", total_revenue)
print("Average revenue per record:", round(average_revenue, 2))
```

## 7. Export your report
Use `to_csv(..., index=False)` to save your cleaned or analysed DataFrame. Create an `output` folder if needed.


```python
output_file = "Student_12_final_analysis.csv"
analysis_df.to_csv(output_file, index=False)

print("Final analysis saved as:", output_file)
```
